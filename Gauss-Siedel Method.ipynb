```python
# Gauss-Siedel Method
```

```python
import numpy as np

# Coefficients of equations
eqns = np.array([[12, 3, -5],
                 [1, 5, 3],
                 [3, 7, 13]])

# Constants
consts = np.array([1, 28, 76])

# Initial guesses
x, y, z = 1, 0.5, 0.6

# Convergence criteria
tol = 1e-6  # Tolerance for stopping criteria
max_iterations = 100  # Maximum iterations

prev_x, prev_y, prev_z = x, y, z

for iteration in range(max_iterations):
    x = (consts[0] - eqns[0][1] * y - eqns[0][2] * z) / eqns[0][0]
    y = (consts[1] - eqns[1][0] * x - eqns[1][2] * z) / eqns[1][1]
    z = (consts[2] - eqns[2][0] * x - eqns[2][1] * y) / eqns[2][2]
    
    print(f"Iteration {iteration + 1}: x = {x:.6f}, y = {y:.6f}, z = {z:.6f}")
    
    # Check for convergence
    if max(abs(x - prev_x), abs(y - prev_y), abs(z - prev_z)) < tol:
        break
    
    prev_x, prev_y, prev_z = x, y, z

print(f"\nFinal solution after convergence:\nx: {x:.6f}\ny: {y:.6f}\nz: {z:.6f}")
```
