# Gauss-Jordan ELimination

```python
import numpy as np

# Coefficients of equations
eqns = np.array([[2, 1, -1, 8],
                 [-3, -1, 2, -11],
                 [-2, 1, 2, -3]], dtype=float)

# Number of equations
n = len(eqns)

# Gauss-Jordan Elimination
for i in range(n):
    eqns[i] = eqns[i] / eqns[i][i]  # Make the diagonal element 1
    for j in range(n):
        if i != j:
            eqns[j] -= eqns[j][i] * eqns[i]  # Make other column elements 0

# Extracting solutions
x, y, z = eqns[:, -1]

# Checking convergence using relative error between iterations
prev_sol = np.zeros(n)
curr_sol = np.array([x, y, z])
tol = 1e-6  # Convergence tolerance
iteration = 0

while np.linalg.norm(curr_sol - prev_sol) > tol:
    prev_sol = curr_sol.copy()
    iteration += 1
    # Reapplying Gauss-Jordan to check convergence over iterations
    for i in range(n):
        eqns[i] = eqns[i] / eqns[i][i]
        for j in range(n):
            if i != j:
                eqns[j] -= eqns[j][i] * eqns[i]
    curr_sol = eqns[:, -1]
    print(f"Iteration {iteration}: x = {curr_sol[0]:.6f}, y = {curr_sol[1]:.6f}, z = {curr_sol[2]:.6f}")

print(f"\nFinal solution after convergence:\nx: {x:.6f}\ny: {y:.6f}\nz: {z:.6f}")
```
